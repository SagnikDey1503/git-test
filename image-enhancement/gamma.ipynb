```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage import img_as_ubyte
```

```python
img = data.camera()
img
```

```python
import numpy as np

def gamma_correction(img, c, g):
    img_flat = img.flatten()
    img_corrected = (c*np.power((img_flat+10e-7) / 255.0, g)) * 255.0
    img_corrected = img_corrected.reshape(img.shape)
    return img_corrected.astype(np.uint8)
```

```python
img_1= gamma_correction(img,1,0.2)
img_2= gamma_correction(img,1,0.4)
img_3= gamma_correction(img,1,0.67)
img_4= gamma_correction(img,0.009,1)
img_5= gamma_correction(img,1,1.5)
img_6= gamma_correction(img,1,2.5)
img_7= gamma_correction(img,1,5)
```

```python
fig, axs = plt.subplots(1, 8, figsize=(18, 5))
axs[0].imshow(img, cmap='gray')
axs[0].set_title("Original Image")
axs[1].imshow(img_1, cmap='gray')
axs[1].set_title("Gamma=0.2")
axs[2].imshow(img_2, cmap='gray')
axs[2].set_title("Gamma=0.4")
axs[3].imshow(img_3, cmap='gray')
axs[3].set_title("Gamma=0.67")
axs[4].imshow(img_4, cmap='gray')
axs[4].set_title("Gamma=1  original and scaled")
axs[5].imshow(img_5, cmap='gray')
axs[5].set_title("Gamma=1.5")
axs[6].imshow(img_6, cmap='gray')
axs[6].set_title("Gamma=2.5")
axs[7].imshow(img_7, cmap='gray')
axs[7].set_title("Gamma=5")

for ax in axs:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
