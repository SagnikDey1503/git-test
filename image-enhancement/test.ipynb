```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage import img_as_ubyte
```

```python
img = data.camera()
img
```

```python
def sigmoid_contrast_stretch(img):
    img = img.astype(np.float32)

    normalized = (img - 127)  #prevent everything from getting too large value
    sigmoid = 1 / (1 + np.exp(-normalized)) 

    stretched = sigmoid * 255

    return stretched.astype(np.uint8)
```

```python
def thresholding(img, threshold=127):
    return np.where(img > threshold, 255, 0).astype(np.uint8)
```

```python
def negative_image(img):
    return 255 - img
```

```python
contrast_img = sigmoid_contrast_stretch(img)
threshold_img = thresholding(img)
negative_img = negative_image(img)
```

```python
fig, axs = plt.subplots(1, 4, figsize=(18, 5))
axs[0].imshow(img, cmap='gray')
axs[0].set_title("Original Image")
axs[1].imshow(contrast_img, cmap='gray')
axs[1].set_title("Contrast Stretching")
axs[2].imshow(threshold_img, cmap='gray')
axs[2].set_title("Thresholding")
axs[3].imshow(negative_img, cmap='gray')
axs[3].set_title("Negative Image")

for ax in axs:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x= np.linspace(0, 255, 10000)-127
y = sigmoid(x)*255

plt.plot(x, y)
plt.title("Standard Sigmoid Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
