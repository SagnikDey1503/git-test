```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

img = data.camera()
img



```

```python
edges = cv2.Canny(img, threshold1=150, threshold2=220)
```

```python
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```
