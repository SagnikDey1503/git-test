```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, color
from skimage.util import random_noise
from skimage.filters import sobel, prewitt, roberts


image = data.camera()
gray = image / 255.0

noisy_image = random_noise(gray, mode='gaussian', var=0.03)

edges_sobel = sobel(gray)
edges_prewitt = prewitt(gray)
edges_roberts = roberts(gray)

edges_sobel_noisy = sobel(noisy_image)
edges_prewitt_noisy = prewitt(noisy_image)
edges_roberts_noisy = roberts(noisy_image)


fig, axs = plt.subplots(2, 4, figsize=(15, 8))

# Original
axs[0, 0].imshow(gray, cmap='gray')
axs[0, 0].set_title('Original')
axs[0, 0].axis('off')

axs[0, 1].imshow(edges_sobel, cmap='gray')
axs[0, 1].set_title('Sobel')
axs[0, 1].axis('off')

axs[0, 2].imshow(edges_prewitt, cmap='gray')
axs[0, 2].set_title('Prewitt')
axs[0, 2].axis('off')

axs[0, 3].imshow(edges_roberts, cmap='gray')
axs[0, 3].set_title('Roberts')
axs[0, 3].axis('off')

# Noisy
axs[1, 0].imshow(noisy_image, cmap='gray')
axs[1, 0].set_title('Noisy Image')
axs[1, 0].axis('off')

axs[1, 1].imshow(edges_sobel_noisy, cmap='gray')
axs[1, 1].set_title('Sobel (noisy)')
axs[1, 1].axis('off')

axs[1, 2].imshow(edges_prewitt_noisy, cmap='gray')
axs[1, 2].set_title('Prewitt (noisy)')
axs[1, 2].axis('off')

axs[1, 3].imshow(edges_roberts_noisy, cmap='gray')
axs[1, 3].set_title('Roberts (noisy)')
axs[1, 3].axis('off')

plt.tight_layout()
plt.show()
```
