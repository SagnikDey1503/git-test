```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

img = data.camera()
img



```

```python
blur1 = cv2.GaussianBlur(img, (5, 5), sigmaX=1.0)
blur2 = cv2.GaussianBlur(img, (5, 5), sigmaX=1.6)

#difference
dog = blur1 - blur2
```

```python

plt.imshow(dog, cmap='gray')
plt.title('DoG - Difference of Gaussians')
plt.axis('off')
plt.show()
```
