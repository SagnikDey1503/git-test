```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, filters
from skimage.segmentation import active_contour


img = io.imread('coin.png')


if img.ndim == 3:
    if img.shape[2] == 4:
        img = color.rgba2rgb(img)
    img = color.rgb2gray(img)

img_smooth = filters.gaussian(img, sigma=2)

height, width = img.shape


s = np.linspace(0, 2*np.pi, 800)
r = (height // 2) + 250 * np.sin(s)
c = (width // 2) + 250 * np.cos(s)
init = np.array([r, c]).T
```

```python

snake = active_contour(
    img_smooth,
    init,
    alpha=0,   # elasticity
    beta=0,       # smoothness
    gamma=0.001,  # step size
    convergence=0.01
)

fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(img, cmap='gray')
ax.plot(init[:, 1], init[:, 0], '--r', label='Initial Contour')
ax.plot(snake[:, 1], snake[:, 0], '-b', linewidth=2, label='Final Contour')
ax.set_title("Active Contour (Snake) Segmentation")
ax.axis('off')
ax.legend()
plt.tight_layout()
plt.show()
```
