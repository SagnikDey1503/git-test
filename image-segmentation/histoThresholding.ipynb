```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread(cv2.samples.findFile('histTest1.png'), cv2.IMREAD_GRAYSCALE)


hist = cv2.calcHist([img], [0], None, [256], [0, 256])
```

```python



threshold_val = 120
# _, thresholded = cv2.threshold(img, threshold_val, 255, cv2.THRESH_BINARY)  //manual
otsu_thresh_val, thresholded = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

print(otsu_thresh_val)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.plot(hist, color='gray')
plt.axvline(threshold_val, color='red', linestyle='--', label=f'Threshold = {threshold_val}')
plt.title("Histogram of Gray Levels")
plt.xlabel("Gray Level")
plt.ylabel("Pixel Count")
plt.legend()


plt.subplot(1, 3, 2)
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(thresholded, cmap='gray')
plt.title("Thresholded Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
