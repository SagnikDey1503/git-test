```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('brain.png', cv2.IMREAD_GRAYSCALE)





seed_point = (img.shape[1] // 2, img.shape[0] // 2)
region_mean = int(img[seed_point[1], seed_point[0]])
threshold = 20

height, width = img.shape
output = np.zeros_like(img, dtype=np.uint8)
visited = np.zeros_like(img, dtype=bool)

stack = [seed_point]
output[seed_point[1], seed_point[0]] = 255
visited[seed_point[1], seed_point[0]] = True


snapshots = []
pixels_grown = 0
save_every =1000 #  before saving a snapshot

while stack:
    x, y = stack.pop()
    for dx in [-1, 0, 1]:
        for dy in [-1, 0, 1]:
            nx, ny = x + dx, y + dy
            if 0 <= nx < width and 0 <= ny < height and not visited[ny, nx]:
                if abs(int(img[ny, nx]) - region_mean) < threshold:
                    output[ny, nx] = 255
                    visited[ny, nx] = True
                    stack.append((nx, ny))
                    pixels_grown += 1
                    if pixels_grown % save_every == 0:
                        snapshots.append(output.copy())

snapshots.append(output)

fig, axs = plt.subplots(1, len(snapshots), figsize=(4 * len(snapshots), 4))
for i, mask in enumerate(snapshots):
    axs[i].imshow(mask, cmap='gray')
    axs[i].set_title(f'Step {i + 1}')
    axs[i].axis('off')
plt.tight_layout()
plt.show()
```
