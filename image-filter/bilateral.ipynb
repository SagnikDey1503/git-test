```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from skimage import data, img_as_ubyte


image = img_as_ubyte(data.camera())


sigma_d_values = [5, 9, 15]
sigma_r_values = [25, 75, 150]


fig, axes = plt.subplots(len(sigma_d_values), len(sigma_r_values), figsize=(12, 9))
fig.suptitle("Bilateral Filtering (cv2.bilateralFilter)", fontsize=16)

for i, sigma_d in enumerate(sigma_d_values):
    for j, sigma_r in enumerate(sigma_r_values):

        filtered = cv2.bilateralFilter(image, d=sigma_d, sigmaColor=sigma_r, sigmaSpace=sigma_d)


        ax = axes[i, j]
        ax.imshow(filtered, cmap='gray')
        ax.set_title(f"σ_d={sigma_d}, σ_r={sigma_r}")
        ax.axis('off')

plt.tight_layout()
plt.subplots_adjust(top=0.92)
plt.show()
```
