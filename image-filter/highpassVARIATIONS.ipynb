```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from scipy.ndimage import convolve

# Load sample grayscale image
image = data.camera()

# Original 3x3 high-pass filter
base_filter = np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
])

# Define fractions from 1/4 to 1/60 (step size can be adjusted)
fractions = [1/i for i in range(1, 200)]  # 1/4 to 1/60

# Plotting
n = len(fractions)
cols = 5
rows = int(np.ceil(n / cols))

fig, axes = plt.subplots(rows, cols, figsize=(15, 3 * rows))

for i, frac in enumerate(fractions):
    scaled_filter = frac * base_filter
    filtered_img = convolve(image, scaled_filter)

    ax = axes[i // cols, i % cols]
    ax.imshow(filtered_img, cmap='gray')
    ax.set_title(f"Scale 1/{int(1/frac)}")
    ax.axis('off')

# Hide unused subplots
for j in range(i + 1, rows * cols):
    axes[j // cols, j % cols].axis('off')

plt.tight_layout()
plt.show()

```
